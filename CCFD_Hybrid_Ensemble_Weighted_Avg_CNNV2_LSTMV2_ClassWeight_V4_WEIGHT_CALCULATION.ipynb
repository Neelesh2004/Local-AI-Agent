```python
# OOAD-structured code
# ============================================
# ENSEMBLE: Modified CNN + Modified LSTM
# Weighted Average Ensemble Strategy
# OOAD Implementation with Class-Based Architecture
# ============================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    precision_recall_curve,
    auc,
    f1_score
)
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import seaborn as sns

print("TensorFlow version:", tf.__version__)


# ============================================
# CLASS 1: DataLoader
# Responsibility: Load and provide access to dataset
# ============================================
class DataLoader:
    """
    Handles data loading operations for the fraud detection system.
    Encapsulates file I/O and initial data validation.
    """

    def __init__(self, file_path):
        self.file_path = file_path
        self.dataframe = None

    def load_data(self):
        """Load dataset from CSV file"""
        self.dataframe = pd.read_csv(self.file_path)
        print("Dataset shape:", self.dataframe.shape)
        return self.dataframe

    def get_class_distribution(self):
        """Display class distribution statistics"""
        print("Original class distribution:")
        print(self.dataframe["Class"].value_counts())
        fraud_percentage = (self.dataframe['Class'].sum() / len(self.dataframe)) * 100
        print(f"Fraud percentage: {fraud_percentage:.4f}%")
        return fraud_percentage


# ============================================
# CLASS 2: DataPreprocessor
# Responsibility: Feature engineering and data preparation
# ============================================
class DataPreprocessor:
    """
    Handles all preprocessing operations including feature scaling,
    train-test split, and data reshaping for different model architectures.
    """

    def __init__(self, dataframe, test_size=0.2, random_state=42):
        self.dataframe = dataframe
        self.test_size = test_size
        self.random_state = random_state
        self.scaler = StandardScaler()

        # Attributes to store processed data
        self.X = None
        self.y = None
        self.X_train = None
        self.X_test = None
        self.y_train = None
        self.y_test = None
        self.X_train_lstm = None
        self.X_test_lstm = None
        self.X_train_cnn = None
        self.X_test_cnn = None

    def scale_features(self):
        """Apply feature scaling to Time and Amount columns"""
        print("\n[Step 1] Feature Scaling for Model Convergence")

        # Separate features and target
        self.X = self.dataframe.drop(["Class"], axis=1)
        self.y = self.dataframe["Class"]

        # Normalize Time
        print("Normalizing 'Time' feature to [0,1] range...")
        self.X["Time"] = (self.X["Time"] - self.X["Time"].min()) / (self.X["Time"].max() - self.X["Time"].min())

        # Standardize Amount
        print("Standardizing 'Amount' feature (Z-score normalization)")
        self.X["Amount"] = self.scaler.fit_transform(self.X["Amount"].values.reshape(-1, 1))

        print("✓ V1-V28 features already PCA-transformed (preserved confidentiality)")
        print("✓ Feature scaling complete")

    def split_data(self):
        """Split data into training and testing sets with stratification"""
        print("\n[Step 2] Dataset Splitting (80% Train / 20% Test)")
        print("Maintaining class distribution via stratification...")

        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, self.y,
            test_size=self.test_size,
            random_state=self.random_state,
            stratify=self.y
        )

        print(f"Training set size: {self.X_train.shape[0]} samples")
        print(f"Test set size: {self.X_test.shape[0]} samples")
        print(f"Training fraud cases: {self.y_train.sum()} ({(self.y_train.sum()/len(self.y_train))*100:.2f}%)")
        print(f"Test fraud cases: {self.y_test.sum()} ({(self.y_test.sum()/len(self.y_test))*100:.2f}%)")

    def reshape_data(self):
        """Reshape data for LSTM and CNN architectures"""
        print("\n[Step 4] Reshaping Data for CNN and LSTM Architectures")

        # For LSTM: (samples, timesteps=1, features=30)
        self.X_train_lstm = self.X_train.values.reshape((self.X_train.shape[0], 1, self.X_train.shape[1]))
        self.X_test_lstm = self.X_test.values.reshape((self.X_test.shape[0], 1, self.X_test.shape[1]))

        # For CNN: (samples, features=30, channels=1)
        self.X_train_cnn = self.X_train.values.reshape((self.X_train.shape[0], self.X_train.shape[1], 1))
        self.X_test_cnn = self.X_test.values.reshape((self.X_test.shape[0], self.X_test.shape[1], 1))

        print(f"LSTM input shape: {self.X_train_lstm.shape} (samples, timesteps, features)")
        print(f"CNN input shape: {self.X_train_cnn.shape} (samples, features, channels)")
        print("✓ Data reshaped for deep learning models")

    def preprocess(self):
        """Execute full preprocessing pipeline"""
        self.scale_features()
        self.split_data()
        self.reshape_data()
        print("\n" + "="*60)
        print("PREPROCESSING COMPLETE - READY FOR MODEL TRAINING")
        print("="*60)


# ============================================
# CLASS 3: ClassWeightHandler
# Responsibility: Manage class imbalance strategy
# ============================================
class ClassWeightHandler:
    """
    Manages class weighting strategy for handling imbalanced datasets.
    Implements FraudX methodology without synthetic sampling.
    """

    def __init__(self, weight_normal=1.0, weight_fraud=2.0):
        self.weight_normal = weight_normal
        self.weight_fraud = weight_fraud
        self.class_weight_dict = {0: weight_normal, 1: weight_fraud}

    def display_strategy(self):
        """Display class imbalance handling strategy"""
        print("\n[Step 3] Handling Class Imbalance with Class Weighting")
        print("NO synthetic sampling (SMOTE)")
        print("Instead: Balanced learning via class weights during training")
        print(f"\nClass weight configuration: {self.class_weight_dict}")
        print("  - Normal transactions (Class 0): weight = 1.0")
        print("  - Fraudulent transactions (Class 1): weight = 2.0")
        print("✓ Higher penalty for misclassifying fraud during training")
        print("✓ Original imbalanced data preserved for real-world applicability")

    def get_weights(self):
        """Return class weight dictionary"""
        return self.class_weight_dict


# ============================================
# CLASS 4: LSTMModelBuilder
# Responsibility: Build and configure LSTM architecture
# ============================================
class LSTMModelBuilder:
    """
    Constructs the Modified LSTM model architecture.
    Encapsulates layer configuration and model compilation.
    """

    def __init__(self, input_shape, learning_rate=0.001):
        self.input_shape = input_shape
        self.learning_rate = learning_rate
        self.model = None

    def build_model(self):
        """Construct LSTM model architecture"""
        print("\n" + "="*60)
        print("BUILDING MODIFIED LSTM MODEL")
        print("="*60)

        self.model = keras.Sequential(name="Modified_LSTM")

        # LSTM Layer 1 (128 units)
        self.model.add(layers.LSTM(128, return_sequences=True, activation='tanh',
                                   input_shape=self.input_shape))
        self.model.add(layers.BatchNormalization())
        self.model.add(layers.Dropout(0.3))

        # LSTM Layer 2 (64 units)
        self.model.add(layers.LSTM(64, return_sequences=True, activation='tanh'))
        self.model.add(layers.BatchNormalization())
        self.model.add(layers.Dropout(0.3))

        # LSTM Layer 3 (32 units)
        self.model.add(layers.LSTM(32, return_sequences=False, activation='tanh'))
        self.model.add(layers.BatchNormalization())
        self.model.add(layers.Dropout(0.3))

        # Dense Layer 1
        self.model.add(layers.Dense(128))
        self.model.add(layers.LeakyReLU(alpha=0.01))
        self.model.add(layers.BatchNormalization())
        self.model.add(layers.Dropout(0.4))

        # Dense Layer 2 (64 units)
        self.model.add(layers.Dense(64))
        self.model.add(layers.LeakyReLU(alpha=0.01))
        self.model.add(layers.BatchNormalization())
        self.model.add(layers.Dropout(0.4))

        # Output layer
        self.model.add(layers.Dense(1, activation='sigmoid'))

        # Compile model
        self.model.compile(
            optimizer=keras.optimizers.Adam(learning_rate=self.learning_rate),
            loss='binary_crossentropy',
            metrics=['accuracy', keras.metrics.AUC(name='auc'), keras.metrics.AUC(name='prc', curve='PR')]
        )

        print("LSTM Model Architecture:")
        self.model.summary()
        return self.model

    def get_model(self):
        """Return the constructed model"""
        return self.model


# ============================================
# CLASS 5: CNNModelBuilder
# Responsibility: Build and configure CNN architecture
# ============================================
class CNNModelBuilder:
    """
    Constructs the Modified CNN model architecture.
    Encapsulates layer configuration and model compilation.
    """

    def __init__(self, input_shape, learning_rate=0.001):
        self.input_shape = input_shape
        self.learning_rate = learning_rate
        self.model = None

    def build_model(self):
        """Construct CNN model architecture"""
        print("\n" + "="*60)
        print("BUILDING MODIFIED CNN MODEL")
        print("="*60)

        self.model = keras.Sequential(name="Modified_CNN")

        # Conv Block 1
        self.model.add(layers.Conv1D(32, 3, padding="same", input_shape=self.input_shape))
        self.model.add(layers.LeakyReLU(alpha=0.01))
        self.model.add(layers.BatchNormalization())
        self.model.add(layers.MaxPooling1D(2))

        # Conv Block 2
        self.model.add(layers.Conv1D(64, 3, padding="same"))
        self.model.add(layers.LeakyReLU(alpha=0.01))
        self.model.add(layers.BatchNormalization())
        self.model.add(layers.MaxPooling1D(2))

        # Conv Block 3
        self.model.add(layers.Conv1D(128, 3, padding="same"))
        self.model.add(layers.LeakyReLU(alpha=0.01))
        self.model.add(layers.BatchNormalization())
        self.model.add(layers.MaxPooling1D(2))

        self.model.add(layers.Flatten())

        # Dense Layers
        self.model.add(layers.Dense(128))
        self.model.add(layers.LeakyReLU(alpha=0.01))
        self.model.add(layers.BatchNormalization())
        self.model.add(layers.Dropout(0.4))

        self.model.add(layers.Dense(64))
        self.model.add(layers.LeakyReLU(alpha=0.01))
        self.model.add(layers.Dropout(0.3))

        # Output layer
        self.model.add(layers.Dense(1, activation="sigmoid"))

        self.model.compile(
            optimizer=keras.optimizers.Adam(learning_rate=self.learning_rate),
            loss="binary_crossentropy",
            metrics=['accuracy', keras.metrics.AUC(name='auc'), keras.metrics.AUC(name='prc', curve='PR')]
        )

        print("CNN Model Architecture:")
        self.model.summary()
        return self.model

    def get_model(self):
        """Return the constructed model"""
        return self.model


# ============================================
# CLASS 6: ModelTrainer
# Responsibility: Train models with callbacks and class weights
# ============================================
class ModelTrainer:
    """
    Handles model training process with callbacks and class weighting.
    Manages training configuration and execution.
    """

    def __init__(self, model, model_name, class_weights):
        self.model = model
        self.model_name = model_name
        self.class_weights = class_weights
        self.history = None

    def train_lstm(self, X_train, y_train, epochs=100, batch_size=128, validation_split=0.2):
        """Train LSTM model"""
        print("\n" + "="*60)
        print("TRAINING LSTM MODEL")
        print("="*60)

        callbacks = [
            keras.callbacks.EarlyStopping(monitor='val_prc', patience=15, mode='max',
                                         restore_best_weights=True, verbose=1),
            keras.callbacks.ReduceLROnPlateau(monitor='val_prc', factor=0.5, patience=5,
                                             mode='max', min_lr=1e-7, verbose=1)
        ]

        self.history = self.model.fit(
            X_train, y_train,
            epochs=epochs,
            batch_size=batch_size,
            validation_split=validation_split,
            callbacks=callbacks,
            class_weight=self.class_weights,
            verbose=1
        )
        return self.history

    def train_cnn(self, X_train, y_train, epochs=50, batch_size=128, validation_split=0.2):
        """Train CNN model"""
        print("\n" + "="*60)
        print("TRAINING CNN MODEL")
        print("="*60)

        callbacks = [
            keras.callbacks.EarlyStopping(monitor="val_prc", patience=10, mode="max",
                                         restore_best_weights=True, verbose=1),
            keras.callbacks.ReduceLROnPlateau(monitor="val_prc", factor=0.5, patience=3,
                                             mode="max", min_lr=1e-7, verbose=1)
        ]

        self.history = self.model.fit(
            X_train, y_train,
            epochs=epochs,
            batch_size=batch_size,
            validation_split=validation_split,
            callbacks=callbacks,
            class_weight=self.class_weights,
            verbose=1
        )
        return self.history


# ============================================
# CLASS 7: ModelPredictor
# Responsibility: Generate predictions from trained models
# ============================================
class ModelPredictor:
    """
    Handles prediction generation from trained models.
    Encapsulates prediction logic for different model types.
    """

    def __init__(self, model):
        self.model = model
        self.predictions_proba = None
        self.predictions_binary = None

    def predict(self, X_test, threshold=0.5):
        """Generate predictions with specified threshold"""
        self.predictions_proba = self.model.predict(X_test).ravel()
        self.predictions_binary = (self.predictions_proba > threshold).astype(int)
        return self.predictions_proba, self.predictions_binary

    def get_probabilities(self):
        """Return probability predictions"""
        return self.predictions_proba

    def get_binary_predictions(self):
        """Return binary predictions"""
        return self.predictions_binary


# ============================================
# CLASS 8: EnsembleWeightCalculator
# Responsibility: Calculate optimal ensemble weights
# ============================================
class EnsembleWeightCalculator:
    """
    Calculates dynamic ensemble weights based on F1-score performance.
    Implements validation-based weight optimization.
    """

    def __init__(self, lstm_model, cnn_model, preprocessor):
        self.lstm_model = lstm_model
        self.cnn_model = cnn_model
        self.preprocessor = preprocessor
        self.weight_lstm = None
        self.weight_cnn = None
        self.lstm_val_f1 = None
        self.cnn_val_f1 = None

    def calculate_weights(self):
        """Calculate F1-based dynamic weights"""
        print("\n" + "="*60)
        print("CREATING WEIGHTED ENSEMBLE WITH F1-BASED DYNAMIC WEIGHTS")
        print("="*60)

        # Create validation set
        X_train_temp, X_val_temp, y_train_temp, y_val_temp = train_test_split(
            self.preprocessor.X_train, self.preprocessor.y_train,
            test_size=0.2, random_state=42, stratify=self.preprocessor.y_train
        )

        # Reshape validation data
        X_val_lstm = X_val_temp.values.reshape((X_val_temp.shape[0], 1, X_val_temp.shape[1]))
        X_val_cnn = X_val_temp.values.reshape((X_val_temp.shape[0], X_val_temp.shape[1], 1))

        print("\n[Step 1] Evaluating models on validation set...")

        # Get validation predictions
        lstm_val_proba = self.lstm_model.predict(X_val_lstm, verbose=0).ravel()
        cnn_val_proba = self.cnn_model.predict(X_val_cnn, verbose=0).ravel()

        lstm_val_pred = (lstm_val_proba > 0.5).astype(int)
        cnn_val_pred = (cnn_val_proba > 0.5).astype(int)

        self.lstm_val_f1 = f1_score(y_val_temp, lstm_val_pred)
        self.cnn_val_f1 = f1_score(y_val_temp, cnn_val_pred)

        print(f"\nValidation F1-Scores:")
        print(f"  • LSTM: {self.lstm_val_f1:.4f}")
        print(f"  • CNN:  {self.cnn_val_f1:.4f}")

        print("\n[Step 2] Calculating weights based on F1-scores...")

        total_f1 = self.lstm_val_f1 + self.cnn_val_f1

        if total_f1 > 0:
            self.weight_lstm = self.lstm_val_f1 / total_f1
            self.weight_cnn = self.cnn_val_f1 / total_f1
        else:
            self.weight_lstm = 0.5
            self.weight_cnn = 0.5
            print("⚠️  Warning: Both models have F1=0, using equal weights")

        print(f"\n✓ Dynamic Weights Calculated:")
        print(f"    • LSTM weight: {self.weight_lstm:.4f} (F1={self.lstm_val_f1:.4f})")
        print(f"    • CNN weight:  {self.weight_cnn:.4f} (F1={self.cnn_val_f1:.4f})")

        if self.weight_lstm > self.weight_cnn:
            ratio = self.weight_lstm / self.weight_cnn
            print(f"\n✓ LSTM is stronger: {ratio:.2f}x more weight than CNN")
        elif self.weight_cnn > self.weight_lstm:
            ratio = self.weight_cnn / self.weight_lstm
            print(f"\n✓ CNN is stronger: {ratio:.2f}x more weight than LSTM")
        else:
            print(f"\n✓ Models are equally weighted")

        return self.weight_lstm, self.weight_cnn

    def get_weights(self):
        """Return calculated weights"""
        return self.weight_lstm, self.weight_cnn


# ============================================
# CLASS 9: EnsemblePredictor
# Responsibility: Generate weighted ensemble predictions
# ============================================
class EnsemblePredictor:
    """
    Creates weighted ensemble predictions by combining multiple model outputs.
    Implements weighted averaging strategy.
    """

    def __init__(self, weight_lstm, weight_cnn):
        self.weight_lstm = weight_lstm
        self.weight_cnn = weight_cnn
        self.ensemble_proba = None
        self.ensemble_pred = None

    def predict(self, lstm_proba, cnn_proba, threshold=0.5):
        """Generate weighted ensemble predictions"""
        print("\n[Step 3] Creating weighted ensemble predictions...")

        self.ensemble_proba = (self.weight_lstm * lstm_proba + self.weight_cnn * cnn_proba)
        self.ensemble_pred = (self.ensemble_proba > threshold).astype(int)

        print("✓ Weighted ensemble predictions created with F1-based dynamic weights")
        print("="*60)

        return self.ensemble_proba, self.ensemble_pred

    def get_predictions(self):
        """Return ensemble predictions"""
        return self.ensemble_proba, self.ensemble_pred


# ============================================
# CLASS 10: ModelEvaluator
# Responsibility: Evaluate model performance
# ============================================
class ModelEvaluator:
    """
    Evaluates model performance using multiple metrics.
    Generates comprehensive evaluation reports.
    """

    def __init__(self, y_true, y_pred, y_pred_proba, model_name):
        self.y_true = y_true
        self.y_pred = y_pred
        self.y_pred_proba = y_pred_proba
        self.model_name = model_name
        self.metrics = {}

    def evaluate(self):
        """Evaluate and print metrics for the model"""
        print(f"\n{'='*60}")
        print(f"{self.model_name} PERFORMANCE")
        print(f"{'='*60}")

        print("\nClassification Report:")
        print(classification_report(self.y_true, self.y_pred, target_names=['Normal', 'Fraud']))

        cm = confusion_matrix(self.y_true, self.y_pred)
        print("\nConfusion Matrix:")
        print(cm)

        tn, fp, fn, tp = cm.ravel()
        accuracy = (tp + tn) / (tp + tn + fp + fn)
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

        roc_auc = roc_auc_score(self.y_true, self.y_pred_proba)
        precision_curve, recall_curve, _ = precision_recall_curve(self.y_true, self.y_pred_proba)
        pr_auc = auc(recall_curve, precision_curve)

        print("\n--- Summary Metrics ---")
        print(f"Accuracy:  {accuracy:.4f}")
        print(f"Precision: {precision:.4f}")
        print(f"Recall:    {recall:.4f}")
        print(f"F1-Score:  {f1:.4f}")
        print(f"ROC-AUC:   {roc_auc:.4f}")
        print(f"PR-AUC:    {pr_auc:.4f}")

        print(f"\n--- Fraud Detection Performance ---")
        print(f"True Positives (Frauds Caught):  {tp}/{fn+tp}")
        print(f"False Negatives (Frauds Missed): {fn}/{fn+tp}")
        print(f"False Positives (False Alarms):  {fp}")
        print(f"True Negatives (Correct Normal): {tn}")

        self.metrics = {
            'accuracy': accuracy,
            'precision': precision,
            'recall': recall,
            'f1': f1,
            'roc_auc': roc_auc,
            'pr_auc': pr_auc,
            'tn': tn, 'fp': fp, 'fn': fn, 'tp': tp
        }

        return self.metrics

    def get_metrics(self):
        """Return evaluation metrics"""
        return self.metrics


# ============================================
# CLASS 11: ResultVisualizer
# Responsibility: Create visualizations for results
# ============================================
class ResultVisualizer:
    """
    Creates visualizations for model comparison and performance analysis.
    Handles confusion matrices and metric comparison plots.
    """

    def __init__(self, y_test, lstm_pred, cnn_pred, ensemble_pred,
                 lstm_metrics, cnn_metrics, ensemble_metrics):
        self.y_test = y_test
        self.lstm_pred = lstm_pred
        self.cnn_pred = cnn_pred
        self.ensemble_pred = ensemble_pred
        self.lstm_metrics = lstm_metrics
        self.cnn_metrics = cnn_metrics
        self.ensemble_metrics = ensemble_metrics

    def plot_confusion_matrices(self):
        """Generate confusion matrix visualizations"""
        print("\n=== Generating Confusion Matrix Visualizations ===")

        fig, axes = plt.subplots(1, 3, figsize=(18, 5))

        cm_lstm = confusion_matrix(self.y_test, self.lstm_pred)
        cm_cnn = confusion_matrix(self.y_test, self.cnn_pred)
        cm_ensemble = confusion_matrix(self.y_test, self.ensemble_pred)

        labels = ['Normal', 'Fraud']

        # Plot 1: LSTM
        sns.heatmap(cm_lstm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=labels, yticklabels=labels,
                    ax=axes[0], cbar=True, annot_kws={"size": 14})
        axes[0].set_title('LSTM Model\nConfusion Matrix', fontweight='bold', fontsize=14)
        axes[0].set_ylabel('Actual Label', fontsize=12)
        axes[0].set_xlabel('Predicted Label', fontsize=12)

        tn, fp, fn, tp = cm_lstm.ravel()
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
        axes[0].text(0.5, -0.15, f'Precision: {precision:.3f} | Recall: {recall:.3f} | F1: {f1:.3f}',
                     ha='center', va='center', transform=axes[0].transAxes, fontsize=11)

        # Plot 2: CNN
        sns.heatmap(cm_cnn, annot=True, fmt='d', cmap='Greens',
                    xticklabels=labels, yticklabels=labels,
                    ax=axes[1], cbar=True, annot_kws={"size": 14})
        axes[1].set_title('CNN Model\nConfusion Matrix', fontweight='bold', fontsize=14)
        axes[1].set_ylabel('Actual Label', fontsize=12)
        axes[1].set_xlabel('Predicted Label', fontsize=12)

        tn, fp, fn, tp = cm_cnn.ravel()
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
        axes[1].text(0.5, -0.15, f'Precision: {precision:.3f} | Recall: {recall:.3f} | F1: {f1:.3f}',
                     ha='center', va='center', transform=axes[1].transAxes, fontsize=11)

        # Plot 3: Ensemble
        sns.heatmap(cm_ensemble, annot=True, fmt='d', cmap='Oranges',
                    xticklabels=labels, yticklabels=labels,
                    ax=axes[2], cbar=True, annot_kws={"size": 14})
        axes[2].set_title('Weighted Ensemble\nConfusion Matrix', fontweight='bold', fontsize=14)
        axes[2].set_ylabel('Actual Label', fontsize=12)
        axes[2].set_xlabel('Predicted Label', fontsize=12)

        tn, fp, fn, tp = cm_ensemble.ravel()
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
        axes[2].text(0.5, -0.15, f'Precision: {precision:.3f} | Recall: {recall:.3f} | F1: {f1:.3f}',
                     ha='center', va='center', transform=axes[2].transAxes, fontsize=11)

        plt.tight_layout()
        plt.savefig('confusion_matrices_comparison.png', dpi=300, bbox_inches='tight')
        plt.show()

        print("✓ Confusion matrices saved as 'confusion_matrices_comparison.png'")

    def plot_metric_comparison(self):
        """Generate metric comparison visualizations"""
        print("\n=== Generating Comparison Visualizations ===")

        fig, axes = plt.subplots(2, 3, figsize=(18, 10))

        models = ['LSTM', 'CNN', 'Weighted\nEnsemble']
        metrics_list = [self.lstm_metrics, self.cnn_metrics, self.ensemble_metrics]
        colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']

        # Plot 1: Precision
        axes[0, 0].bar(models, [m['precision'] for m in metrics_list], color=colors)
        axes[0, 0].set_title('Precision Comparison', fontweight='bold', fontsize=12)
        axes[0, 0].set_ylabel('Precision', fontsize=10)
        axes[0, 0].set_ylim([0, 1])
        axes[0, 0].grid(axis='y', alpha=0.3)

        # Plot 2: Recall
        axes[0, 1].bar(models, [m['recall'] for m in metrics_list], color=colors)
        axes[0, 1].set_title('Recall Comparison', fontweight='bold', fontsize=12)
        axes[0, 1].set_ylabel('Recall', fontsize=10)
        axes[0, 1].set_ylim([0, 1])
        axes[0, 1].grid(axis='y', alpha=0.3)

        # Plot 3: F1-Score
        axes[0, 2].bar(models, [m['f1'] for m in metrics_list], color=colors)
        axes[0, 2].set_title('F1-Score Comparison', fontweight='bold', fontsize=12)
        axes[0, 2].set_ylabel('F1-Score', fontsize=10)
        axes[0, 2].set_ylim([0, 1])
        axes[0, 2].grid(axis='y', alpha=0.3)

        # Plot 4: ROC-AUC
        axes[1, 0].bar(models, [m['roc_auc'] for m in metrics_list], color=colors)
        axes[1, 0].set_title('ROC-AUC Comparison', fontweight='bold', fontsize=12)
        axes[1, 0].set_ylabel('ROC-AUC', fontsize=10)
        axes[1, 0].set_ylim([0.9, 1.0])
        axes[1, 0].grid(axis='y', alpha=0.3)

        # Plot 5: False Positives
        axes[1, 1].bar(models, [m['fp'] for m in metrics_list], color=colors)
        axes[1, 1].set_title('False Positives Comparison', fontweight='bold', fontsize=12)
        axes[1, 1].set_ylabel('Count', fontsize=10)
        axes[1, 1].grid(axis='y', alpha=0.3)

        # Plot 6: False Negatives
        axes[1, 2].bar(models, [m['fn'] for m in metrics_list], color=colors)
        axes[1, 2].set_title('False Negatives Comparison', fontweight='bold', fontsize=12)
        axes[1, 2].set_ylabel('Count', fontsize=10)
        axes[1, 2].grid(axis='y', alpha=0.3)

        plt.tight_layout()
        plt.savefig('weighted_ensemble_comparison.png', dpi=300, bbox_inches='tight')
        plt.show()

        print("✓ Visualization saved as 'weighted_ensemble_comparison.png'")


# ============================================
# CLASS 12: ResultReporter
# Responsibility: Generate comprehensive comparison reports
# ============================================
class ResultReporter:
    """
    Generates comprehensive comparison tables and reports.
    Consolidates results from multiple models.
    """

    def __init__(self, lstm_metrics, cnn_metrics, ensemble_metrics):
        self.lstm_metrics = lstm_metrics
        self.cnn_metrics = cnn_metrics
        self.ensemble_metrics = ensemble_metrics

    def generate_comparison_table(self):
        """Generate and display comparison table"""
        print("\n" + "="*80)
        print("COMPREHENSIVE COMPARISON: LSTM vs CNN vs WEIGHTED ENSEMBLE")
        print("="*80)

        comparison_df = pd.DataFrame({
            'Model': ['LSTM', 'CNN', 'Weighted Ensemble'],
            'Accuracy': [self.lstm_metrics['accuracy'], self.cnn_metrics['accuracy'],
                        self.ensemble_metrics['accuracy']],
            'Precision': [self.lstm_metrics['precision'], self.cnn_metrics['precision'],
                         self.ensemble_metrics['precision']],
            'Recall': [self.lstm_metrics['recall'], self.cnn_metrics['recall'],
                      self.ensemble_metrics['recall']],
            'F1-Score': [self.lstm_metrics['f1'], self.cnn_metrics['f1'],
                        self.ensemble_metrics['f1']],
            'ROC-AUC': [self.lstm_metrics['roc_auc'], self.cnn_metrics['roc_auc'],
                       self.ensemble_metrics['roc_auc']],
            'PR-AUC': [self.lstm_metrics['pr_auc'], self.cnn_metrics['pr_auc'],
                      self.ensemble_metrics['pr_auc']],
            'False Positives': [self.lstm_metrics['fp'], self.cnn_metrics['fp'],
                               self.ensemble_metrics['fp']],
            'False Negatives': [self.lstm_metrics['fn'], self.cnn_metrics['fn'],
                               self.ensemble_metrics['fn']]
        })

        print("\n" + comparison_df.to_string(index=False))

        print("\n" + "="*80)
        print("BEST PERFORMING MODEL PER METRIC")
        print("="*80)
        print(f"Best Accuracy:  {comparison_df.loc[comparison_df['Accuracy'].idxmax(), 'Model']} ({comparison_df['Accuracy'].max():.4f})")
        print(f"Best Precision: {comparison_df.loc[comparison_df['Precision'].idxmax(), 'Model']} ({comparison_df['Precision'].max():.4f})")
        print(f"Best Recall:    {comparison_df.loc[comparison_df['Recall'].idxmax(), 'Model']} ({comparison_df['Recall'].max():.4f})")
        print(f"Best F1-Score:  {comparison_df.loc[comparison_df['F1-Score'].idxmax(), 'Model']} ({comparison_df['F1-Score'].max():.4f})")
        print(f"Best ROC-AUC:   {comparison_df.loc[comparison_df['ROC-AUC'].idxmax(), 'Model']} ({comparison_df['ROC-AUC'].max():.4f})")
        print(f"Best PR-AUC:    {comparison_df.loc[comparison_df['PR-AUC'].idxmax(), 'Model']} ({comparison_df['PR-AUC'].max():.4f})")

        return comparison_df


# ============================================
# CLASS 13: ModelPersistence
# Responsibility: Save and load models and predictions
# ============================================
class ModelPersistence:
    """
    Handles model and prediction persistence operations.
    Manages file I/O for trained models and results.
    """

    @staticmethod
    def save_models(lstm_model, cnn_model, ensemble_proba):
        """Save trained models and ensemble predictions"""
        lstm_model.save('ensemble_lstm.h5')
        cnn_model.save('ensemble_cnn.h5')
        np.save('ensemble_weighted_predictions.npy', ensemble_proba)

        print("\n✓ LSTM model saved as 'ensemble_lstm.h5'")
        print("✓ CNN model saved as 'ensemble_cnn.h5'")
        print("✓ Ensemble predictions saved as 'ensemble_weighted_predictions.npy'")


# ============================================
# CLASS 14: FraudDetectionSystem (Main Controller)
# Responsibility: Orchestrate entire fraud detection pipeline
# ============================================
class FraudDetectionSystem:
    """
    Main controller class that orchestrates the entire fraud detection pipeline.
    Coordinates all components following OOAD principles.
    """

    def __init__(self, data_file_path):
        self.data_file_path = data_file_path

        # Component objects
        self.data_loader = None
        self.preprocessor = None
        self.class_weight_handler = None
        self.lstm_builder = None
        self.cnn_builder = None
        self.lstm_trainer = None
        self.cnn_trainer = None
        self.lstm_predictor = None
        self.cnn_predictor = None
        self.weight_calculator = None
        self.ensemble_predictor = None
        self.visualizer = None
        self.reporter = None

        # Data and models
        self.dataframe = None
        self.lstm_model = None
        self.cnn_model = None

        # Results
        self.lstm_metrics = None
        self.cnn_metrics = None
        self.ensemble_metrics = None
        self.weight_lstm = None
        self.weight_cnn = None

    def load_data(self):
        """Step 1: Load dataset"""
        print("\n" + "="*80)
        print("STEP 1: DATA LOADING")
        print("="*80)

        self.data_loader = DataLoader(self.data_file_path)
        self.dataframe = self.data_loader.load_data()
        self.data_loader.get_class_distribution()

    def preprocess_data(self):
        """Step 2: Preprocess data"""
        print("\n" + "="*80)
        print("STEP 2: DATA PREPROCESSING")
        print("="*80)

        self.preprocessor = DataPreprocessor(self.dataframe)
        self.preprocessor.preprocess()

        self.class_weight_handler = ClassWeightHandler()
        self.class_weight_handler.display_strategy()

    def build_models(self):
        """Step 3: Build model architectures"""
        print("\n" + "="*80)
        print("STEP 3: MODEL ARCHITECTURE CONSTRUCTION")
        print("="*80)

        # Build LSTM
        lstm_input_shape = (self.preprocessor.X_train_lstm.shape[1],
                           self.preprocessor.X_train_lstm.shape[2])
        self.lstm_builder = LSTMModelBuilder(lstm_input_shape)
        self.lstm_model = self.lstm_builder.build_model()

        # Build CNN
        cnn_input_shape = (self.preprocessor.X_train_cnn.shape[1], 1)
        self.cnn_builder = CNNModelBuilder(cnn_input_shape)
        self.cnn_model = self.cnn_builder.build_model()

    def train_models(self):
        """Step 4: Train models"""
        print("\n" + "="*80)
        print("STEP 4: MODEL TRAINING")
        print("="*80)

        class_weights = self.class_weight_handler.get_weights()

        # Train LSTM
        self.lstm_trainer = ModelTrainer(self.lstm_model, "LSTM", class_weights)
        self.lstm_trainer.train_lstm(self.preprocessor.X_train_lstm,
                                     self.preprocessor.y_train)

        # Train CNN
        self.cnn_trainer = ModelTrainer(self.cnn_model, "CNN", class_weights)
        self.cnn_trainer.train_cnn(self.preprocessor.X_train_cnn,
                                   self.preprocessor.y_train)

    def generate_predictions(self):
        """Step 5: Generate individual model predictions"""
        print("\n" + "="*80)
        print("STEP 5: PREDICTION GENERATION")
        print("="*80)
        print("GENERATING PREDICTIONS ON IMBALANCED TEST SET")

        # LSTM predictions
        self.lstm_predictor = ModelPredictor(self.lstm_model)
        lstm_proba, lstm_pred = self.lstm_predictor.predict(self.preprocessor.X_test_lstm)
        print("✓ LSTM predictions generated")

        # CNN predictions
        self.cnn_predictor = ModelPredictor(self.cnn_model)
        cnn_proba, cnn_pred = self.cnn_predictor.predict(self.preprocessor.X_test_cnn)
        print("✓ CNN predictions generated")

        return lstm_proba, lstm_pred, cnn_proba, cnn_pred

    def create_ensemble(self, lstm_proba, cnn_proba):
        """Step 6: Create weighted ensemble"""
        print("\n" + "="*80)
        print("STEP 6: ENSEMBLE CREATION")
        print("="*80)

        # Calculate weights
        self.weight_calculator = EnsembleWeightCalculator(self.lstm_model,
                                                          self.cnn_model,
                                                          self.preprocessor)
        self.weight_lstm, self.weight_cnn = self.weight_calculator.calculate_weights()

        # Generate ensemble predictions
        self.ensemble_predictor = EnsemblePredictor(self.weight_lstm, self.weight_cnn)
        ensemble_proba, ensemble_pred = self.ensemble_predictor.predict(lstm_proba, cnn_proba)

        return ensemble_proba, ensemble_pred

    def evaluate_models(self, lstm_pred, lstm_proba, cnn_pred, cnn_proba,
                       ensemble_pred, ensemble_proba):
        """Step 7: Evaluate all models"""
        print("\n" + "="*80)
        print("STEP 7: MODEL EVALUATION")
        print("="*80)

        # Evaluate LSTM
        lstm_evaluator = ModelEvaluator(self.preprocessor.y_test, lstm_pred,
                                       lstm_proba, "LSTM MODEL")
        self.lstm_metrics = lstm_evaluator.evaluate()

        # Evaluate CNN
        cnn_evaluator = ModelEvaluator(self.preprocessor.y_test, cnn_pred,
                                      cnn_proba, "CNN MODEL")
        self.cnn_metrics = cnn_evaluator.evaluate()

        # Evaluate Ensemble
        ensemble_evaluator = ModelEvaluator(self.preprocessor.y_test, ensemble_pred,
                                           ensemble_proba, "WEIGHTED ENSEMBLE (LSTM + CNN)")
        self.ensemble_metrics = ensemble_evaluator.evaluate()

    def visualize_results(self, lstm_pred, cnn_pred, ensemble_pred):
        """Step 8: Generate visualizations"""
        print("\n" + "="*80)
        print("STEP 8: RESULT VISUALIZATION")
        print("="*80)

        self.visualizer = ResultVisualizer(self.preprocessor.y_test,
                                          lstm_pred, cnn_pred, ensemble_pred,
                                          self.lstm_metrics, self.cnn_metrics,
                                          self.ensemble_metrics)
        self.visualizer.plot_confusion_matrices()
        self.visualizer.plot_metric_comparison()

    def generate_report(self):
        """Step 9: Generate comparison report"""
        print("\n" + "="*80)
        print("STEP 9: COMPREHENSIVE REPORTING")
        print("="*80)

        self.reporter = ResultReporter(self.lstm_metrics, self.cnn_metrics,
                                      self.ensemble_metrics)
        self.reporter.generate_comparison_table()

    def save_results(self, ensemble_proba):
        """Step 10: Save models and predictions"""
        print("\n" + "="*80)
        print("STEP 10: MODEL PERSISTENCE")
        print("="*80)

        ModelPersistence.save_models(self.lstm_model, self.cnn_model, ensemble_proba)

    def print_summary(self):
        """Print final summary"""
        print("\n" + "="*80)
        print("PREPROCESSING METHODOLOGY SUMMARY")
        print("="*80)
        print(f"\n1. DATASET:")
        print(f"   - Total transactions: {len(self.dataframe)}")
        print(f"   - Features: 31 (28 PCA-transformed + Time + Amount + Class)")
        print(f"   - Class imbalance: {(self.dataframe['Class'].sum()/len(self.dataframe))*100:.4f}% fraud")

        print("\n2. FEATURE ENGINEERING:")
        print("   - V1-V28: PCA-transformed (anonymized for confidentiality)")
        print("   - Time: Min-max normalized to [0,1]")
        print("   - Amount: Standardized (Z-score normalization)")

        print("\n3. DATASET SPLITTING:")
        print(f"   - Training: {len(self.preprocessor.X_train)} samples (80%)")
        print(f"   - Testing: {len(self.preprocessor.X_test)} samples (20%)")
        print("   - Stratified split maintains class distribution")

        print("\n4. CLASS IMBALANCE HANDLING (FraudX Strategy):")
        print("   ✓ NO synthetic sampling (SMOTE not used)")
        print("   ✓ Balanced learning via class weights: {0:1.0, 1:2.0}")
        print("   ✓ 2x penalty for misclassifying fraud transactions")
        print("   ✓ Original imbalanced data preserved")

        print("\n5. EVALUATION STRATEGY:")
        print("   - Models trained on imbalanced data with class weights")
        print("   - Evaluated on original imbalanced test set")
        print("   - Ensures real-world operational applicability")

        print("\n6. ENSEMBLE APPROACH:")
        print("   - Weighted average of LSTM and CNN predictions")
        print(f"   - Weights: LSTM={self.weight_lstm:.4f}, CNN={self.weight_cnn:.4f}")
        print("   - Combines strengths of both architectures")

        print("\n" + "="*80)
        print("WEIGHTED ENSEMBLE ANALYSIS COMPLETE")
        print("="*80)
        print("✓ LSTM model trained and saved")
        print("✓ CNN model trained and saved")
        print("✓ Weighted ensemble predictions generated")
        print("✓ Comprehensive comparison completed")
        print("✓ Visualizations saved")
        print(f"✓ Ensemble weights: LSTM={self.weight_lstm:.4f}, CNN={self.weight_cnn:.4f}")
        print("="*80)

    def run(self):
        """Execute complete fraud detection pipeline"""
        print("\n" + "="*80)
        print("FRAUD DETECTION SYSTEM")
        print("Object-Oriented Architecture Implementation")
        print("="*80)

        # Execute pipeline
        self.load_data()
        self.preprocess_data()
        self.build_models()
        self.train_models()

        lstm_proba, lstm_pred, cnn_proba, cnn_pred = self.generate_predictions()
        ensemble_proba, ensemble_pred = self.create_ensemble(lstm_proba, cnn_proba)

        self.evaluate_models(lstm_pred, lstm_proba, cnn_pred, cnn_proba,
                           ensemble_pred, ensemble_proba)
        self.visualize_results(lstm_pred, cnn_pred, ensemble_pred)
        self.generate_report()
        self.save_results(ensemble_proba)
        self.print_summary()


# ============================================
# MAIN EXECUTION
# ============================================
if __name__ == "__main__":
    # Initialize and run the fraud detection system
    fraud_system = FraudDetectionSystem("/content/drive/MyDrive/CCDF_European_Dataset/creditcard.csv")
    fraud_system.run()

print("OOAD-Structured Code Generated Successfully!")
print("\n" + "="*80)
print("CODE STRUCTURE SUMMARY")
print("="*80)
print("\nTotal Classes: 14")
print("\n1. DataLoader - Handles data loading operations")
print("2. DataPreprocessor - Feature engineering and data preparation")
print("3. ClassWeightHandler - Manages class imbalance strategy")
print("4. LSTMModelBuilder - Constructs LSTM architecture")
print("5. CNNModelBuilder - Constructs CNN architecture")
print("6. ModelTrainer - Trains models with callbacks")
print("7. ModelPredictor - Generates predictions")
print("8. EnsembleWeightCalculator - Calculates ensemble weights")
print("9. EnsemblePredictor - Creates weighted ensemble predictions")
print("10. ModelEvaluator - Evaluates model performance")
print("11. ResultVisualizer - Creates visualizations")
print("12. ResultReporter - Generates comparison reports")
print("13. ModelPersistence - Saves/loads models")
print("14. FraudDetectionSystem - Main controller (orchestrates pipeline)")
print("\n" + "="*80)
```

```python

# ============================================
# CREDIT CARD FRAUD DETECTION WITH EXPLAINABLE AI
# OOAD Implementation with Class-Based Architecture
# ============================================
!pip install lime
from lime.lime_tabular import LimeTabularExplainer
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
from tensorflow import keras
import shap
import warnings
warnings.filterwarnings('ignore')


# ============================================
# CLASS 1: DataManager
# Responsibility: Load and manage dataset and models
# ============================================
class DataManager:
    """
    Handles data loading, model loading, and initial preprocessing.
    Encapsulates file I/O operations and data validation.
    """

    def __init__(self, data_path, lstm_model_path, cnn_model_path):
        self.data_path = data_path
        self.lstm_model_path = lstm_model_path
        self.cnn_model_path = cnn_model_path

        # Data attributes
        self.dataframe = None
        self.X = None
        self.y = None
        self.X_train = None
        self.X_test = None
        self.y_train = None
        self.y_test = None

        # Model attributes
        self.lstm_model = None
        self.cnn_model = None

        # Preprocessing attributes
        self.scaler = StandardScaler()
        self.time_min = None
        self.time_max = None

    def load_data(self):
        """Load dataset from CSV file"""
        print("\n[Step 1] Loading models and data...")
        self.dataframe = pd.read_csv(self.data_path)
        print(f"✓ Dataset loaded: {self.dataframe.shape[0]} transactions")
        return self.dataframe

    def load_models(self):
        """Load pre-trained LSTM and CNN models"""
        self.lstm_model = keras.models.load_model(self.lstm_model_path)
        self.cnn_model = keras.models.load_model(self.cnn_model_path)
        print("✓ Models loaded successfully!")
        return self.lstm_model, self.cnn_model

    def preprocess_data(self):
        """Apply preprocessing transformations to dataset"""
        # Separate features and target
        self.X = self.dataframe.drop(["Class"], axis=1)
        self.y = self.dataframe["Class"]

        # Store min/max for Time normalization
        self.time_min = self.X["Time"].min()
        self.time_max = self.X["Time"].max()

        # Normalize Time
        self.X["Time"] = (self.X["Time"] - self.time_min) / (self.time_max - self.time_min)

        # Standardize Amount
        self.X["Amount"] = self.scaler.fit_transform(self.X["Amount"].values.reshape(-1, 1))

    def split_data(self, test_size=0.2, random_state=42):
        """Split data into training and testing sets"""
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, self.y, test_size=test_size, random_state=random_state, stratify=self.y
        )
        print(f"✓ Data split: {self.X_train.shape[0]} train, {self.X_test.shape[0]} test")

    def initialize(self):
        """Execute full initialization pipeline"""
        self.load_data()
        self.load_models()
        self.preprocess_data()
        self.split_data()
        return self

    def get_models(self):
        """Return loaded models"""
        return self.lstm_model, self.cnn_model

    def get_data(self):
        """Return processed data splits"""
        return self.X_train, self.X_test, self.y_train, self.y_test

    def get_scaler(self):
        """Return fitted scaler for new data preprocessing"""
        return self.scaler

    def get_time_bounds(self):
        """Return Time feature bounds for normalization"""
        return self.time_min, self.time_max


# ============================================
# CLASS 2: EnsembleWeightCalculator
# Responsibility: Calculate dynamic ensemble weights
# ============================================
class EnsembleWeightCalculator:
    """
    Calculates optimal ensemble weights based on validation performance.
    Implements F1-score based weight optimization strategy.
    """

    def __init__(self, lstm_model, cnn_model, X_train, y_train):
        self.lstm_model = lstm_model
        self.cnn_model = cnn_model
        self.X_train = X_train
        self.y_train = y_train

        self.weight_lstm = None
        self.weight_cnn = None
        self.lstm_val_f1 = None
        self.cnn_val_f1 = None

    def calculate_weights(self, val_size=0.2, random_state=42):
        """Calculate F1-based dynamic ensemble weights"""
        print("\n[Step 2] Calculating dynamic ensemble weights...")

        # Create validation set
        X_train_temp, X_val_temp, y_train_temp, y_val_temp = train_test_split(
            self.X_train, self.y_train,
            test_size=val_size, random_state=random_state, stratify=self.y_train
        )

        # Reshape validation data
        X_val_lstm = X_val_temp.values.reshape((X_val_temp.shape[0], 1, X_val_temp.shape[1]))
        X_val_cnn = X_val_temp.values.reshape((X_val_temp.shape[0], X_val_temp.shape[1], 1))

        # Get validation predictions
        lstm_val_proba = self.lstm_model.predict(X_val_lstm, verbose=0).ravel()
        cnn_val_proba = self.cnn_model.predict(X_val_cnn, verbose=0).ravel()

        # Calculate F1-scores
        lstm_val_pred = (lstm_val_proba > 0.5).astype(int)
        cnn_val_pred = (cnn_val_proba > 0.5).astype(int)

        self.lstm_val_f1 = f1_score(y_val_temp, lstm_val_pred)
        self.cnn_val_f1 = f1_score(y_val_temp, cnn_val_pred)

        # Calculate weights
        total_f1 = self.lstm_val_f1 + self.cnn_val_f1

        if total_f1 > 0:
            self.weight_lstm = self.lstm_val_f1 / total_f1
            self.weight_cnn = self.cnn_val_f1 / total_f1
        else:
            self.weight_lstm = 0.5
            self.weight_cnn = 0.5
            print("⚠️  Warning: Both models have F1=0, using equal weights")

        print(f"\n✓ Dynamic Weights Calculated:")
        print(f"    • LSTM weight: {self.weight_lstm:.4f} (Validation F1={self.lstm_val_f1:.4f})")
        print(f"    • CNN weight:  {self.weight_cnn:.4f} (Validation F1={self.cnn_val_f1:.4f})")

        return self.weight_lstm, self.weight_cnn

    def get_weights(self):
        """Return calculated weights"""
        return self.weight_lstm, self.weight_cnn


# ============================================
# CLASS 3: ModelPredictionWrapper
# Responsibility: Wrap models for SHAP compatibility
# ============================================
class ModelPredictionWrapper:
    """
    Provides wrapper functions for model predictions compatible with SHAP.
    Handles data reshaping for different model architectures.
    """

    def __init__(self, lstm_model, cnn_model, weight_lstm, weight_cnn):
        self.lstm_model = lstm_model
        self.cnn_model = cnn_model
        self.weight_lstm = weight_lstm
        self.weight_cnn = weight_cnn

    def predict_lstm(self, data):
        """Wrapper for LSTM predictions with automatic reshaping"""
        data_reshaped = data.reshape((data.shape[0], 1, 30))
        return self.lstm_model.predict(data_reshaped, verbose=0)

    def predict_cnn(self, data):
        """Wrapper for CNN predictions with automatic reshaping"""
        data_reshaped = data.reshape((data.shape[0], 30, 1))
        return self.cnn_model.predict(data_reshaped, verbose=0)

    def predict_ensemble(self, data):
        """Wrapper for ensemble predictions with dynamic weights"""
        lstm_pred = self.predict_lstm(data)
        cnn_pred = self.predict_cnn(data)
        return self.weight_lstm * lstm_pred + self.weight_cnn * cnn_pred


# ============================================
# CLASS 4: SHAPExplainerManager
# Responsibility: Initialize and manage SHAP explainers
# ============================================
class SHAPExplainerManager:
    """
    Manages SHAP explainer initialization and configuration.
    Creates background datasets and explainer instances.
    """

    def __init__(self, prediction_wrapper, X_train, background_size=50, random_state=42):
        self.prediction_wrapper = prediction_wrapper
        self.X_train = X_train
        self.background_size = background_size
        self.random_state = random_state

        self.background_data = None
        self.explainer_lstm = None
        self.explainer_cnn = None
        self.explainer_ensemble = None

    def initialize_explainers(self):
        """Initialize SHAP explainers for all models"""
        print("\n[Step 3] Initializing SHAP explainers...")

        # Create background dataset
        np.random.seed(self.random_state)
        background_indices = np.random.choice(
            self.X_train.shape[0], self.background_size, replace=False
        )
        self.background_data = self.X_train.iloc[background_indices].values

        # Initialize SHAP explainers
        self.explainer_lstm = shap.KernelExplainer(
            self.prediction_wrapper.predict_lstm, self.background_data
        )
        self.explainer_cnn = shap.KernelExplainer(
            self.prediction_wrapper.predict_cnn, self.background_data
        )
        self.explainer_ensemble = shap.KernelExplainer(
            self.prediction_wrapper.predict_ensemble, self.background_data
        )

        print("✓ SHAP explainers ready!")
        return self

    def get_explainers(self):
        """Return initialized explainers"""
        return self.explainer_lstm, self.explainer_cnn, self.explainer_ensemble

# ============================================
# CLASS 4b: LIMEExplainerManager
# Responsibility: Initialize and manage LIME explainer
# ============================================
class LIMEExplainerManager:
    """
    Manages LIME explainer initialization and configuration for tabular data.
    Uses the ensemble prediction wrapper for local explanations.
    """

    def __init__(self, prediction_wrapper, X_train):
        self.prediction_wrapper = prediction_wrapper
        self.X_train = X_train
        self.feature_names = ['Time'] + [f'V{i}' for i in range(1, 29)] + ['Amount']
        self.explainer_lime = None

    def initialize_explainer(self):
        """
        Initialize LIME Tabular Explainer.
        Uses training data as background distribution.
        """
        print("\n[Step 3b] Initializing LIME explainer...")

        self.explainer_lime = LimeTabularExplainer(
            training_data=self.X_train.values,
            feature_names=self.feature_names,
            class_names=['LEGITIMATE', 'FRAUD'],
            mode='classification'  # binary classification
        )
        print("✓ LIME explainer ready!")
        return self

    def get_explainer(self):
        """Return initialized LIME explainer"""
        return self.explainer_lime


# ============================================
# CLASS 5: TransactionSelector
# Responsibility: Handle transaction selection from test set or manual input
# ============================================
class TransactionSelector:
    """
    Manages transaction selection for testing.
    Handles fraud/legitimate sampling and manual input validation.
    """

    def __init__(self, X_test, y_test):
        self.X_test = X_test
        self.y_test = y_test

    def display_menu(self):
        """Display selection menu to user"""
        print("\n" + "="*80)
        print("TEST OPTIONS:")
        print("="*80)
        print("1. Test FRAUD transaction (with classification + XAI)")
        print("2. Test LEGITIMATE transaction (with classification + XAI)")
        print("3. Enter custom transaction manually (with classification + XAI)")

    def get_user_choice(self):
        """Get and validate user choice"""
        choice = input("\n👉 Choose (1/2/3): ").strip()
        return choice

    def select_fraud_transaction(self):
        """Select random fraud transaction from test set"""
        fraud_indices = self.y_test[self.y_test == 1].index.tolist()
        fraud_idx = np.random.choice(fraud_indices)
        transaction = self.X_test.loc[fraud_idx].values
        test_position = self.X_test.index.get_loc(fraud_idx)

        print(f"\n✓ Selected FRAUD transaction from test set (Position: {test_position})")
        return transaction, "FRAUD", test_position

    def select_legitimate_transaction(self):
        """Select random legitimate transaction from test set"""
        legit_indices = self.y_test[self.y_test == 0].index.tolist()
        legit_idx = np.random.choice(legit_indices)
        transaction = self.X_test.loc[legit_idx].values
        test_position = self.X_test.index.get_loc(legit_idx)

        print(f"\n✓ Selected LEGITIMATE transaction from test set (Position: {test_position})")
        return transaction, "LEGITIMATE", test_position

    def get_manual_input(self):
        """Get manual transaction input from user"""
        print("\n" + "="*80)
        print("MANUAL TRANSACTION INPUT")
        print("="*80)
        print("\nEnter 30 comma-separated values in this order:")
        print("Time, V1, V2, V3, ..., V28, Amount")
        print("\nExample:")
        print("45123.5, -1.2345, 0.5678, ..., 25.50")

        user_input = input("\n👉 Enter values: ").strip()
        return user_input


# ============================================
# CLASS 6: TransactionPreprocessor
# Responsibility: Preprocess manual input transactions
# ============================================
class TransactionPreprocessor:
    """
    Handles preprocessing of manual transaction inputs.
    Applies same transformations as training data.
    """

    def __init__(self, scaler, time_min, time_max):
        self.scaler = scaler
        self.time_min = time_min
        self.time_max = time_max

    def preprocess_manual_input(self, raw_values):
        """Preprocess manual transaction input"""
        # Create feature names
        feature_names = ['Time'] + [f'V{i}' for i in range(1, 29)] + ['Amount']
        input_df = pd.DataFrame([raw_values], columns=feature_names)

        print("\n" + "="*80)
        print("📋 YOUR INPUT (Before Preprocessing)")
        print("="*80)
        print(f"\nTime (original):   {raw_values[0]:.2f} seconds")
        print(f"Amount (original): ${raw_values[29]:.2f}")

        # Apply preprocessing
        input_df['Time'] = (input_df['Time'] - self.time_min) / (self.time_max - self.time_min)
        input_df['Amount'] = self.scaler.transform(input_df['Amount'].values.reshape(-1, 1))

        transaction_preprocessed = input_df.values[0]

        print(f"\nTime (normalized):    {transaction_preprocessed[0]:.6f}")
        print(f"Amount (standardized): {transaction_preprocessed[29]:.6f}")
        print("="*80)

        return transaction_preprocessed


# ============================================
# CLASS 7: TransactionClassifier
# Responsibility: Classify transactions using ensemble model
# ============================================
class TransactionClassifier:
    """
    Performs transaction classification using the weighted ensemble.
    Generates predictions from individual models and combines them.
    """

    def __init__(self, lstm_model, cnn_model, weight_lstm, weight_cnn):
        self.lstm_model = lstm_model
        self.cnn_model = cnn_model
        self.weight_lstm = weight_lstm
        self.weight_cnn = weight_cnn

    def classify_transaction(self, transaction_data):
        """Classify single transaction and return predictions"""
        print("\n Processing through ensemble model...")

        # Reshape for models
        input_lstm = transaction_data.reshape((1, 1, 30))
        input_cnn = transaction_data.reshape((1, 30, 1))

        # Predict
        lstm_prob = self.lstm_model.predict(input_lstm, verbose=0)[0][0]
        cnn_prob = self.cnn_model.predict(input_cnn, verbose=0)[0][0]
        ensemble_prob = (self.weight_lstm * lstm_prob + self.weight_cnn * cnn_prob)

        return lstm_prob, cnn_prob, ensemble_prob

    def get_prediction_label(self, ensemble_prob):
        """Convert probability to prediction label"""
        return "FRAUD" if ensemble_prob > 0.5 else "LEGITIMATE"


# ============================================
# CLASS 8: ResultDisplayer
# Responsibility: Display classification results
# ============================================
class ResultDisplayer:
    """
    Handles display of transaction data and classification results.
    Formats output for user comprehension.
    """

    def __init__(self, weight_lstm, weight_cnn):
        self.weight_lstm = weight_lstm
        self.weight_cnn = weight_cnn

    def display_transaction_data(self, transaction_data, test_index=None):
        """Display transaction feature values"""
        print("\n" + "="*80)
        if test_index is not None:
            print(f" TRANSACTION DATA (Test Index: {test_index})")
        else:
            print(" TRANSACTION DATA (Custom Input)")
        print("="*80)
        print(f"\nTime (normalized):    {transaction_data[0]:.6f}")
        print(f"Amount (standardized): {transaction_data[29]:.6f}")
        print(f"\nPCA Features V1-V28:")
        print("-"*80)
        for i in range(1, 29):
            print(f"  V{i:2d}: {transaction_data[i]:8.6f}", end="")
            if i % 4 == 0:
                print()
        print("\n" + "="*80)

    def display_classification_results(self, lstm_prob, cnn_prob, ensemble_prob,
                                      actual_label=None):
        """Display classification results and predictions"""
        print("\n" + "="*80)
        print(" CLASSIFICATION RESULTS")
        print("="*80)
        print(f"\nIndividual Model Predictions:")
        print(f"  • LSTM:     {lstm_prob:.4f} ({lstm_prob*100:.2f}%)")
        print(f"  • CNN:      {cnn_prob:.4f} ({cnn_prob*100:.2f}%)")
        print(f"\nWeighted Ensemble:")
        print(f"  • Weights:  LSTM={self.weight_lstm:.3f}, CNN={self.weight_cnn:.3f}")
        print(f"  • Ensemble: {ensemble_prob:.4f} ({ensemble_prob*100:.2f}%)")
        print(f"\n FINAL PREDICTION: {'FRAUD ' if ensemble_prob > 0.5 else 'LEGITIMATE '}")

        if actual_label and actual_label != "UNKNOWN":
            print(f"\n📌 Ground Truth: {actual_label}")
            correct = ((ensemble_prob > 0.5 and actual_label == "FRAUD") or
                      (ensemble_prob <= 0.5 and actual_label == "LEGITIMATE"))
            print(f"✓ Prediction Match: {'YES ✅' if correct else 'NO ❌'}")

        if ensemble_prob > 0.5:
            print("\n HIGH RISK - RECOMMEND MANUAL REVIEW")
            print("    Action: Flag for fraud investigation team")
        else:
            print("\n LOW RISK - TRANSACTION APPROVED")
            print("    Action: Proceed with transaction")

        print("="*80)


# ============================================
# CLASS 9: SHAPValueCalculator
# Responsibility: Calculate SHAP values for explanations
# ============================================
class SHAPValueCalculator:
    """
    Computes SHAP values for transaction explanations.
    Handles calculation for multiple models.
    """

    def __init__(self, explainer_lstm, explainer_cnn, explainer_ensemble):
        self.explainer_lstm = explainer_lstm
        self.explainer_cnn = explainer_cnn
        self.explainer_ensemble = explainer_ensemble

    def calculate_shap_values(self, transaction_data, nsamples=100):
        """Calculate SHAP values for all models"""
        print("\n" + "="*80)
        print("🔍 GENERATING EXPLAINABLE AI (XAI) ANALYSIS")
        print("="*80)

        transaction_reshaped = transaction_data.reshape(1, -1)

        print(f"\n Computing SHAP values (nsamples={nsamples})...")
        print("   This analyzes feature contributions using Shapley values...")
        print("   (Estimated time: 30-60 seconds)")

        print("\n   [1/3] Computing Ensemble SHAP values...")
        shap_values_ensemble = self.explainer_ensemble.shap_values(
            transaction_reshaped, nsamples=nsamples
        )

        print("   [2/3] Computing LSTM SHAP values...")
        shap_values_lstm = self.explainer_lstm.shap_values(
            transaction_reshaped, nsamples=nsamples
        )

        print("   [3/3] Computing CNN SHAP values...")
        shap_values_cnn = self.explainer_cnn.shap_values(
            transaction_reshaped, nsamples=nsamples
        )

        return shap_values_lstm, shap_values_cnn, shap_values_ensemble

# ============================================
# CLASS 9b: LIMEValueCalculator
# Responsibility: Calculate LIME explanation for a transaction
# ============================================
class LIMEValueCalculator:
    """
    Computes LIME explanations for a single transaction.
    Uses the ensemble prediction function for local surrogate fitting.
    """

    def __init__(self, lime_explainer, prediction_wrapper):
        self.lime_explainer = lime_explainer
        self.prediction_wrapper = prediction_wrapper

    def _predict_proba_ensemble(self, data_np):
        """
        LIME requires a predict_proba-like function:
        input: (n_samples, n_features)
        output: (n_samples, 2) probabilities for [LEGITIMATE, FRAUD]
        """
        # Get ensemble probability of fraud (shape (n_samples, 1))
        fraud_prob = self.prediction_wrapper.predict_ensemble(data_np).ravel()
        # Convert to 2-class probabilities
        legit_prob = 1.0 - fraud_prob
        probs = np.vstack([legit_prob, fraud_prob]).T
        return probs

    def explain_transaction(self, transaction_data, num_features=10, num_samples=5000):
        """
        Generate LIME explanation for a single transaction
        Returns the LIME explanation object.
        """
        print("\n   [LIME] Generating local explanation...")
        exp = self.lime_explainer.explain_instance(
            data_row=transaction_data,
            predict_fn=self._predict_proba_ensemble,
            num_features=num_features,
            top_labels=1,
            num_samples=num_samples
        )
        return exp


# ============================================
# CLASS 10: SHAPVisualizationGenerator
# Responsibility: Generate SHAP visualizations
# ============================================
class SHAPVisualizationGenerator:
    """
    Creates SHAP-based visualizations for model explanations.
    Generates waterfall plots and feature importance charts.
    """

    def __init__(self, explainer_lstm, explainer_cnn, weight_lstm, weight_cnn):
        self.explainer_lstm = explainer_lstm
        self.explainer_cnn = explainer_cnn
        self.weight_lstm = weight_lstm
        self.weight_cnn = weight_cnn
        self.feature_names = ['Time'] + [f'V{i}' for i in range(1, 29)] + ['Amount']

    def generate_visualizations(self, transaction_data, shap_values_lstm,
                               shap_values_cnn, shap_values_ensemble, test_index=None):
        """Generate comprehensive SHAP visualizations"""
        print("\n Generating SHAP visualizations...")

        fig, axes = plt.subplots(2, 2, figsize=(20, 12))
        display_index = test_index if test_index is not None else "Custom"
        fig.suptitle(
            f'SHAP Explanations - Transaction {display_index}\n'
            f'(Dynamic Weighted Ensemble: LSTM={self.weight_lstm:.3f}, CNN={self.weight_cnn:.3f})',
            fontsize=16, fontweight='bold'
        )

        # Flatten SHAP values
        shap_lstm_flat = np.array(shap_values_lstm).flatten()
        shap_cnn_flat = np.array(shap_values_cnn).flatten()
        shap_ensemble_flat = np.array(shap_values_ensemble).flatten()

        # Get base values
        base_value_lstm = self._extract_base_value(self.explainer_lstm.expected_value)
        base_value_cnn = self._extract_base_value(self.explainer_cnn.expected_value)

        # Generate plots
        self._plot_lstm_waterfall(axes[0, 0], shap_lstm_flat, base_value_lstm, transaction_data)
        self._plot_cnn_waterfall(axes[0, 1], shap_cnn_flat, base_value_cnn, transaction_data)
        self._plot_lstm_bar(axes[1, 0], shap_lstm_flat)
        self._plot_cnn_bar(axes[1, 1], shap_cnn_flat)

        plt.tight_layout()
        save_name = f'shap_explanation_{display_index}.png'
        plt.savefig(save_name, dpi=300, bbox_inches='tight')
        plt.show()

        print(f"\n✓ Visualization saved as '{save_name}'")
        return save_name, shap_ensemble_flat

    def _extract_base_value(self, expected_value):
        """Extract base value from SHAP explainer"""
        if isinstance(expected_value, np.ndarray):
            return float(expected_value[0])
        else:
            return float(expected_value)

    def _plot_lstm_waterfall(self, ax, shap_values, base_value, transaction_data):
        """Plot LSTM waterfall chart"""
        plt.sca(ax)
        shap.waterfall_plot(
            shap.Explanation(values=shap_values,
                           base_values=base_value,
                           data=transaction_data,
                           feature_names=self.feature_names),
            max_display=15,
            show=False
        )
        ax.set_title(f'LSTM Model (weight={self.weight_lstm:.3f})\nFeature Impact Waterfall',
                     fontweight='bold', fontsize=12)

    def _plot_cnn_waterfall(self, ax, shap_values, base_value, transaction_data):
        """Plot CNN waterfall chart"""
        plt.sca(ax)
        shap.waterfall_plot(
            shap.Explanation(values=shap_values,
                           base_values=base_value,
                           data=transaction_data,
                           feature_names=self.feature_names),
            max_display=15,
            show=False
        )
        ax.set_title(f'CNN Model (weight={self.weight_cnn:.3f})\nFeature Impact Waterfall',
                     fontweight='bold', fontsize=12)

    def _plot_lstm_bar(self, ax, shap_values):
        """Plot LSTM feature importance bar chart"""
        plt.sca(ax)
        shap_abs = np.abs(shap_values)
        top_indices = np.argsort(shap_abs)[-10:][::-1]

        colors = ['#FF4444' if shap_values[i] > 0 else '#4444FF' for i in top_indices]
        plt.barh([self.feature_names[i] for i in top_indices],
                 [shap_values[i] for i in top_indices],
                 color=colors, alpha=0.7, edgecolor='black', linewidth=0.5)
        plt.xlabel('SHAP Value (Impact on Prediction)', fontsize=11, fontweight='bold')
        plt.title('LSTM - Top 10 Most Influential Features', fontweight='bold', fontsize=12)
        plt.axvline(x=0, color='black', linestyle='--', linewidth=1)
        plt.grid(axis='x', alpha=0.3)

    def _plot_cnn_bar(self, ax, shap_values):
        """Plot CNN feature importance bar chart"""
        plt.sca(ax)
        shap_abs = np.abs(shap_values)
        top_indices = np.argsort(shap_abs)[-10:][::-1]

        colors = ['#FF4444' if shap_values[i] > 0 else '#4444FF' for i in top_indices]
        plt.barh([self.feature_names[i] for i in top_indices],
                 [shap_values[i] for i in top_indices],
                 color=colors, alpha=0.7, edgecolor='black', linewidth=0.5)
        plt.xlabel('SHAP Value (Impact on Prediction)', fontsize=11, fontweight='bold')
        plt.title('CNN - Top 10 Most Influential Features', fontweight='bold', fontsize=12)
        plt.axvline(x=0, color='black', linestyle='--', linewidth=1)
        plt.grid(axis='x', alpha=0.3)


# ============================================
# CLASS 10b: LIMEVisualizationGenerator
# Responsibility: Visualize / print LIME explanations
# ============================================
class LIMEVisualizationGenerator:
    def __init__(self):
        self.feature_names = ['Time'] + [f'V{i}' for i in range(1, 29)] + ['Amount']

    def display_lime_explanation(self, lime_exp):
        print("\n" + "="*80)
        print(" LIME LOCAL EXPLANATION (Ensemble Model)")
        print("="*80)

        predicted_label = lime_exp.available_labels()[0]
        exp_list = lime_exp.as_list(label=predicted_label)

        print(f"\nPredicted label index (LIME): {predicted_label}")
        print(f"\n{'Rank':<6}{'Feature':<30}{'Contribution':<15}")
        print("-"*60)
        for i, (feat_str, contrib) in enumerate(exp_list, 1):
            print(f"{i:<6}{feat_str:<30}{contrib:+.6f}")  # <-- fixed

        print("\nNote: Positive contribution pushes toward FRAUD, "
              "negative toward LEGITIMATE (based on class mapping).")
        print("="*80)

    def plot_lime_explanation(self, lime_exp):
        """
        Optional: generate a matplotlib figure summarizing LIME explanation.
        """
        fig = lime_exp.as_pyplot_figure()
        plt.tight_layout()
        plt.show()


# ============================================
# CLASS 11: FeatureContributionAnalyzer
# Responsibility: Analyze and display feature contributions
# ============================================
class FeatureContributionAnalyzer:
    """
    Analyzes SHAP values to identify top contributing features.
    Generates feature contribution tables and interpretations.
    """

    def __init__(self):
        self.feature_names = ['Time'] + [f'V{i}' for i in range(1, 29)] + ['Amount']

    def display_top_features(self, shap_values, transaction_data, top_n=10):
        """Display top N feature contributions"""
        print("\n" + "="*80)
        print(f" TOP {top_n} FEATURE CONTRIBUTIONS (Ensemble Model)")
        print("="*80)

        top_indices = np.argsort(np.abs(shap_values))[-top_n:][::-1]

        print(f"\n{'Rank':<6}{'Feature':<12}{'SHAP Value':<18}{'Impact':<18}{'Feature Value':<15}")
        print("-"*85)
        for rank, idx in enumerate(top_indices, 1):
            impact = "→ FRAUD (↑)" if shap_values[idx] > 0 else "→ LEGIT (↓)"
            print(f"{rank:<6}{self.feature_names[idx]:<12}{shap_values[idx]:>+.6f}{'':10}"
                  f"{impact:<18}{transaction_data[idx]:>+.6f}")

        self._display_interpretation_guide()

    def _display_interpretation_guide(self):
        """Display interpretation guide for SHAP values"""
        print("\n" + "="*80)
        print("INTERPRETATION GUIDE:")
        print("="*80)
        print("• Positive SHAP values (red bars) → Push prediction TOWARD fraud")
        print("• Negative SHAP values (blue bars) → Push prediction TOWARD legitimate")
        print("• Larger absolute values → Stronger influence on prediction")
        print("• SHAP values sum to (final prediction - base value)")
        print("="*80)


# ============================================
# CLASS 12: ExplainableClassificationPipeline
# Responsibility: Orchestrate classification + explanation workflow
# ============================================
class ExplainableClassificationPipeline:
    def __init__(self, classifier, result_displayer, shap_calculator,
                 visualization_generator, contribution_analyzer,
                 lime_calculator, lime_visualizer):
        self.classifier = classifier
        self.result_displayer = result_displayer
        self.shap_calculator = shap_calculator
        self.visualization_generator = visualization_generator
        self.contribution_analyzer = contribution_analyzer
        self.lime_calculator = lime_calculator
        self.lime_visualizer = lime_visualizer

    def classify_and_explain(self, transaction_data, actual_label="UNKNOWN",
                            test_index=None, nsamples=100):
        """
        Execute complete classification and explanation pipeline

        Parameters:
        -----------
        transaction_data : array
            30 feature values
        actual_label : str
            "FRAUD", "LEGITIMATE", or "UNKNOWN"
        test_index : int
            Optional test set index for display
        nsamples : int
            SHAP sampling parameter
        """
        # Step 1: Display transaction data
        self.result_displayer.display_transaction_data(transaction_data, test_index)

        # Step 2: Classify transaction
        lstm_prob, cnn_prob, ensemble_prob = self.classifier.classify_transaction(
            transaction_data
        )

        # Step 3: Display results
        self.result_displayer.display_classification_results(
            lstm_prob, cnn_prob, ensemble_prob, actual_label
        )

        # Step 4: Calculate SHAP values
        shap_lstm, shap_cnn, shap_ensemble = self.shap_calculator.calculate_shap_values(
            transaction_data, nsamples
        )

        # Step 4b: Calculate LIME explanation (local surrogate)
        lime_exp = self.lime_calculator.explain_transaction(
            transaction_data, num_features=10, num_samples=5000
        )

        # Step 5: Generate SHAP visualizations
        save_name, shap_ensemble_flat = self.visualization_generator.generate_visualizations(
            transaction_data, shap_lstm, shap_cnn, shap_ensemble, test_index
        )

        # Step 5b: Display LIME explanation
        self.lime_visualizer.display_lime_explanation(lime_exp)
        self.lime_visualizer.plot_lime_explanation(lime_exp)

        # Step 6: Display feature contributions (SHAP-based)
        self.contribution_analyzer.display_top_features(shap_ensemble_flat, transaction_data)


        print("\n✓ Classification and XAI explanation complete!\n")


# ============================================
# CLASS 13: FraudDetectionXAISystem (Main Controller)
# Responsibility: Orchestrate entire XAI system
# ============================================
class FraudDetectionXAISystem:
    """
    Main controller class for the fraud detection XAI system.
    Coordinates all components and manages user interaction workflow.
    """

    def __init__(self, data_path, lstm_model_path, cnn_model_path):
        self.data_path = data_path
        self.lstm_model_path = lstm_model_path
        self.cnn_model_path = cnn_model_path

        # Component instances
        self.data_manager = None
        self.weight_calculator = None
        self.prediction_wrapper = None
        self.explainer_manager = None
        self.transaction_selector = None
        self.transaction_preprocessor = None
        self.classifier = None
        self.result_displayer = None
        self.shap_calculator = None
        self.visualization_generator = None
        self.contribution_analyzer = None
        self.lime_explainer_manager = None
        self.lime_calculator = None
        self.lime_visualizer = None

        self.pipeline = None

        # System attributes
        self.weight_lstm = None
        self.weight_cnn = None

    def initialize_system(self):
        """Initialize all system components"""
        print("\n" + "="*80)
        print("🔍 CREDIT CARD FRAUD DETECTION WITH EXPLAINABLE AI")
        print("="*80)

        # Step 1: Initialize data manager
        self.data_manager = DataManager(
            self.data_path, self.lstm_model_path, self.cnn_model_path
        )
        self.data_manager.initialize()

        lstm_model, cnn_model = self.data_manager.get_models()
        X_train, X_test, y_train, y_test = self.data_manager.get_data()

        # Step 2: Calculate ensemble weights
        self.weight_calculator = EnsembleWeightCalculator(
            lstm_model, cnn_model, X_train, y_train
        )
        self.weight_lstm, self.weight_cnn = self.weight_calculator.calculate_weights()

        # Step 3: Initialize prediction wrapper
        self.prediction_wrapper = ModelPredictionWrapper(
            lstm_model, cnn_model, self.weight_lstm, self.weight_cnn
        )

        # Step 4: Initialize SHAP explainer manager
        self.explainer_manager = SHAPExplainerManager(
            self.prediction_wrapper, X_train
        )
        self.explainer_manager.initialize_explainers()

        explainer_lstm, explainer_cnn, explainer_ensemble = (
            self.explainer_manager.get_explainers()
        )

        # Step 4b: Initialize LIME explainer manager
        self.lime_explainer_manager = LIMEExplainerManager(
            self.prediction_wrapper, X_train
        )
        self.lime_explainer_manager.initialize_explainer()
        lime_explainer = self.lime_explainer_manager.get_explainer()


        # Step 5: Initialize other components
        self.transaction_selector = TransactionSelector(X_test, y_test)
        self.transaction_preprocessor = TransactionPreprocessor(
            self.data_manager.get_scaler(),
            *self.data_manager.get_time_bounds()
        )
        self.classifier = TransactionClassifier(
            lstm_model, cnn_model, self.weight_lstm, self.weight_cnn
        )
        self.result_displayer = ResultDisplayer(self.weight_lstm, self.weight_cnn)
        self.shap_calculator = SHAPValueCalculator(
            explainer_lstm, explainer_cnn, explainer_ensemble
        )
        self.visualization_generator = SHAPVisualizationGenerator(
            explainer_lstm, explainer_cnn, self.weight_lstm, self.weight_cnn
        )
        self.contribution_analyzer = FeatureContributionAnalyzer()
        self.lime_calculator = LIMEValueCalculator(
            lime_explainer, self.prediction_wrapper
        )
        self.lime_visualizer = LIMEVisualizationGenerator()


        # Step 6: Initialize main pipeline
        self.pipeline = ExplainableClassificationPipeline(
            self.classifier,
            self.result_displayer,
            self.shap_calculator,
            self.visualization_generator,
            self.contribution_analyzer,
            self.lime_calculator,
            self.lime_visualizer
        )

    def handle_user_interaction(self):
        """Handle user choice and execute appropriate workflow"""
        self.transaction_selector.display_menu()
        choice = self.transaction_selector.get_user_choice()

        if choice == '1':
            self._handle_fraud_transaction()
        elif choice == '2':
            self._handle_legitimate_transaction()
        elif choice == '3':
            self._handle_manual_input()
        else:
            print("❌ Invalid choice. Please run again and select 1, 2, or 3.")

    def _handle_fraud_transaction(self):
        """Handle fraud transaction selection and analysis"""
        transaction, label, test_idx = self.transaction_selector.select_fraud_transaction()
        self.pipeline.classify_and_explain(transaction, label, test_idx, nsamples=100)

    def _handle_legitimate_transaction(self):
        """Handle legitimate transaction selection and analysis"""
        transaction, label, test_idx = self.transaction_selector.select_legitimate_transaction()
        self.pipeline.classify_and_explain(transaction, label, test_idx, nsamples=100)

    def _handle_manual_input(self):
        """Handle manual transaction input and analysis"""
        user_input = self.transaction_selector.get_manual_input()

        try:
            transaction = [float(x.strip()) for x in user_input.split(',')]
            if len(transaction) != 30:
                print(f"❌ Error: Expected 30 values, got {len(transaction)}")
                print("   Please ensure you provide: Time + 28 V-features + Amount")
            else:
                transaction_preprocessed = self.transaction_preprocessor.preprocess_manual_input(
                    transaction
                )
                self.pipeline.classify_and_explain(
                    transaction_preprocessed, "UNKNOWN", nsamples=100
                )
        except ValueError as e:
            print(f"❌ Error: Invalid input format")
            print(f"   Details: {e}")
            print("   Ensure all values are numbers separated by commas")
        except Exception as e:
            print(f"❌ Unexpected error: {e}")

    def display_system_summary(self):
        """Display final system configuration summary"""
        print("\n" + "="*80)
        print(" FRAUD DETECTION WITH EXPLAINABLE AI - SESSION COMPLETE")
        print("="*80)
        print("\n📊 System Configuration:")
        print(f"   ✓ Classification: CNN-LSTM Dynamic Weighted Ensemble")
        print(f"   ✓ Ensemble Weights: LSTM={self.weight_lstm:.4f}, CNN={self.weight_cnn:.4f}")
        print(f"   ✓ Weight Calculation: F1-Score based (validation set)")
        print(f"   ✓ Explainability: SHAP (KernelExplainer) + LIME (Tabular Explainer)")
        print(f"   ✓ Background Samples: 50 transactions")
        print(f"   ✓ SHAP Samples: 100 feature combinations")
        print("="*80)

    def run(self):
        """Execute complete fraud detection XAI system"""
        self.initialize_system()
        self.handle_user_interaction()
        self.display_system_summary()


# ============================================
# MAIN EXECUTION
# ============================================
if __name__ == "__main__":
    # Initialize and run the XAI system
    xai_system = FraudDetectionXAISystem(
        data_path="/content/drive/MyDrive/CCDF_European_Dataset/creditcard.csv",
        lstm_model_path='ensemble_lstm.h5',
        cnn_model_path='ensemble_cnn.h5'
    )
    xai_system.run()

print("OOAD-Structured XAI Code Generated Successfully!")
print("\n" + "="*80)
print("CODE STRUCTURE SUMMARY - XAI MODULE")
print("="*80)
print("\nTotal Classes: 13")
print("\n1. DataManager - Loads data, models, and handles preprocessing")
print("2. EnsembleWeightCalculator - Calculates dynamic ensemble weights")
print("3. ModelPredictionWrapper - Wraps models for SHAP compatibility")
print("4. SHAPExplainerManager - Initializes and manages SHAP explainers")
print("5. TransactionSelector - Handles transaction selection from test set")
print("6. TransactionPreprocessor - Preprocesses manual input transactions")
print("7. TransactionClassifier - Classifies transactions using ensemble")
print("8. ResultDisplayer - Displays classification results")
print("9. SHAPValueCalculator - Calculates SHAP values for explanations")
print("10. SHAPVisualizationGenerator - Generates SHAP visualizations")
print("11. FeatureContributionAnalyzer - Analyzes feature contributions")
print("12. ExplainableClassificationPipeline - Orchestrates classify + explain")
print("13. FraudDetectionXAISystem - Main controller for entire XAI system")
print("\n" + "="*80)
print("\nKEY OOAD PRINCIPLES DEMONSTRATED:")
print("="*80)
print("✓ Single Responsibility Principle - Each class has one clear purpose")
print("✓ Separation of Concerns - Logic divided into cohesive modules")
print("✓ Encapsulation - Data and behavior bundled within classes")
print("✓ Composition over Inheritance - Components work together")
print("✓ High Cohesion - Related functionality grouped together")
print("✓ Low Coupling - Classes interact through clean interfaces")
print("="*80)
```
